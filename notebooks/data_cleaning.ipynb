```python
import pandas as pd

# load required csv files
customers = pd.read_csv('olist_customers_dataset.csv')
order_items = pd.read_csv('olist_order_items_dataset.csv')
orders = pd.read_csv('olist_orders_dataset.csv')
payments = pd.read_csv('olist_order_payments_dataset.csv')
products = pd.read_csv('olist_products_dataset.csv')
```

```python

# select required columns
customers = customers[[
    "customer_id",
    "customer_unique_id",
    "customer_city",
    "customer_state"
]]

order_items = order_items[[
    "order_id",
    "order_item_id",
    "product_id",
    "seller_id",
    "shipping_limit_date",
    "price",
    "freight_value"
]]

orders = orders[[
    "order_id",
    "customer_id",
    "order_status",
    "order_purchase_timestamp",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]]

payments = payments[[
    "order_id",
    "payment_sequential",
    "payment_type",
    "payment_installments",
    "payment_value"
]]

products = products[[
    "product_id",
    "product_category_name"
]]
```

remove null values in orders

```python
na_values = ["na","n/a",""," ","NA","N/A"]

def clean_dataset(df):

  for col in df.columns:
    if df[col].dtype == 'object':
      df[col] = df[col].str.strip()
      df[col] = df[col].str.lower()
      df[col] = df[col].replace(na_values,np.nan)

  return df
```

```python
customers = clean_dataset(customers)
order_items = clean_dataset(order_items)
orders = clean_dataset(orders)
payments = clean_dataset(payments)
products = clean_dataset(products)
```

```python
order_items.isnull().sum()
```

```python
orders.isnull().sum()
```

```python

# 1. Incomplete orders ko remove karo (~ operator use karke)
orders = orders[
    ~orders['order_status'].isin(['canceled', 'unavailable', 'invoiced', 'processing'])
]

# 2. Ab valid orders me se agar koi null delivery date hai toh use drop karo
orders = orders.dropna(
    subset=['order_delivered_customer_date']
)

# 3. Final null values check karo
orders.isnull().sum()
```

fill null values in product category

```python
products['product_category_name'] = products['product_category_name'].fillna('unknown')
products.isnull().sum()
```

```python
orders['order_purchase_timestamp'] = pd.to_datetime(orders['order_purchase_timestamp'], dayfirst=True).dt.floor('D')
orders['order_delivered_customer_date'] = pd.to_datetime(orders['order_delivered_customer_date'], dayfirst=True).dt.floor('D')
orders['order_estimated_delivery_date'] = pd.to_datetime(orders['order_estimated_delivery_date'], dayfirst=True).dt.floor('D')
order_items['shipping_limit_date']  = pd.to_datetime(order_items['shipping_limit_date'],dayfirst=True).dt.floor('D')
```
